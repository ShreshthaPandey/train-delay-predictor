```python
st_zone = stations[["station_name", "station_zone"]].drop_duplicates("station_name")

sz = schedule.merge(st_zone, on="station_name", how="left")
sz["is_nr"] = sz["station_zone"] == "NR"

share = (sz.groupby("train_no")["is_nr"]
           .agg(nr_share="mean", stops="count")
           .reset_index())

print(share["nr_share"].describe())
print("Trains with 70%+ stops in NR:", (share["nr_share"] >= 0.7).sum())
print("Trains with 50%+ stops in NR:", (share["nr_share"] >= 0.5).sum())
```

```python
nr_trains = share.loc[share["nr_share"] >= 0.7, "train_no"].tolist()
print(len(nr_trains), nr_trains[:5])

nr_df = pd.DataFrame({"train_no": nr_trains})
con.register("nr_trains", nr_df)

con.sql("""
SELECT COUNT(*) AS rows,
       COUNT(DISTINCT d.train_no) AS trains,
       COUNT(DISTINCT d.station_name) AS stations
FROM 'railway_data/delay_final.parquet' d
JOIN nr_trains n ON d.train_no = n.train_no
""").show()
```

```python
con.sql("""
COPY (
  SELECT d.*
  FROM 'railway_data/delay_final.parquet' d
  JOIN nr_trains n ON d.train_no = n.train_no
  ORDER BY d.train_no, d.date, d.station_no
) TO 'railway_data/delay_nr.parquet' (FORMAT PARQUET)
""")
```

```python
df = pd.read_parquet("railway_data/delay_nr.parquet")
df["date"] = pd.to_datetime(df["date"])
df["train_no"] = df["train_no"].astype("category")
df["station_name"] = df["station_name"].astype("category")

print(df.shape)
print(df.dtypes)
print(df.head())
```

```python
# 1. schedule ko sirf apni trains tak chhota karo
sched = schedule[schedule["train_no"].isin(nr_trains)][
    ["train_no", "station_no", "station_name",
     "distance_from_origin", "arrival_day", "arrival_time"]
].drop_duplicates(["train_no", "station_no", "station_name"]).copy()

# 2. df jaisi category type do (merge tez aur kam RAM)
sched["train_no"] = pd.Categorical(sched["train_no"], categories=df["train_no"].cat.categories)
sched["station_name"] = pd.Categorical(sched["station_name"], categories=df["station_name"].cat.categories)

# 3. train details taiyar karo
tr_info = trains[["train_no", "train_name", "type_code"]].drop_duplicates("train_no").copy()
tr_info["train_no"] = pd.Categorical(tr_info["train_no"], categories=df["train_no"].cat.categories)

# 4. merge
df = df.merge(sched, on=["train_no", "station_no", "station_name"], how="left")
df = df.merge(tr_info, on="train_no", how="left")

print(df.shape)
print(df.isna().mean().round(4))
```

```python
miss = df["distance_from_origin"].isna()

# 1. Kitni trains affected hain, aur kitni poori tarah unmatched?
per_train = miss.groupby(df["train_no"], observed=True).mean()
print("Trains with any miss:", (per_train > 0).sum())
print("Trains fully missing (>99%):", (per_train > 0.99).sum())
print(per_train.describe())

# 2. Kya station sahi hai par station_no alag hai?
alt = df.loc[miss, ["train_no", "station_name"]].drop_duplicates()
alt = alt.merge(sched[["train_no", "station_name"]].drop_duplicates(),
                on=["train_no", "station_name"], how="left", indicator=True)
print(alt["_merge"].value_counts())
```

```python
df = df.sort_values(["train_no", "date", "station_no"]).reset_index(drop=True)

df["prev_delay"] = df.groupby(["train_no", "date"], observed=True)["delay"].shift(1)

print(df[["date", "train_no", "station_name", "station_no", "delay", "prev_delay"]].head(8))
print("prev_delay NaN %:", round(df["prev_delay"].isna().mean() * 100, 2))
print("Correlation delay vs prev_delay:", round(df["delay"].corr(df["prev_delay"]), 3))
```

```python
df["dayofweek"] = df["date"].dt.dayofweek
df["month"] = df["date"].dt.month
df["is_fog_season"] = df["month"].isin([12, 1]).astype(int)

# arrival_time "11:00" jaisa text hai, uska ghanta nikalo
df["sched_hour"] = pd.to_numeric(df["arrival_time"].astype(str).str[:2], errors="coerce")

print(df[["date", "dayofweek", "month", "is_fog_season", "arrival_time", "sched_hour"]].head(8))
print(df[["dayofweek", "month", "is_fog_season", "sched_hour"]].isna().mean().round(4))
```

```python
df = df.drop(columns=["distance_from_origin", "arrival_day",
                      "arrival_time", "sched_hour"])

df = df.merge(
    sched[["train_no", "station_no", "station_name",
           "distance_from_origin", "arrival_day", "arrival_time"]],
    on=["train_no", "station_no", "station_name"], how="left")

df["sched_hour"] = pd.to_numeric(df["arrival_time"].astype(str).str[:2], errors="coerce")

print(df.shape)
print(df[["distance_from_origin", "arrival_day", "arrival_time", "sched_hour"]].isna().mean().round(4))
```

```python
g = df.groupby(["train_no", "date"], observed=True)

df["stop_idx"] = g.cumcount()
df["route_stops"] = g["station_no"].transform("count")
df["stops_left"] = df["route_stops"] - df["stop_idx"] - 1
df["route_progress"] = df["stop_idx"] / (df["route_stops"] - 1).clip(lower=1)

print(df[["train_no", "date", "station_name", "stop_idx",
          "route_stops", "stops_left", "route_progress"]].head(8))
print(df[["stop_idx", "route_stops", "stops_left", "route_progress"]].isna().sum())
```

```python
# 1. Har run ka average delay
runs = (df.groupby(["train_no", "date"], observed=True)["delay"]
          .mean().reset_index(name="run_mean")
          .sort_values(["train_no", "date"]))

# 2. Pichle 7 runs ka average (aaj ki run chhodkar)
runs["avg_last7"] = (runs.groupby("train_no", observed=True)["run_mean"]
                         .transform(lambda s: s.shift(1).rolling(7, min_periods=3).mean()))

# 3. df me jodo
df = df.merge(runs[["train_no", "date", "avg_last7"]],
              on=["train_no", "date"], how="left")

print(df.shape)
print("avg_last7 NaN %:", round(df["avg_last7"].isna().mean() * 100, 2))
print("Correlation delay vs avg_last7:", round(df["delay"].corr(df["avg_last7"]), 3))
print(df[["train_no", "date", "delay", "avg_last7"]].drop_duplicates(["train_no", "date"]).head(8))
```

```python
from sklearn.metrics import mean_absolute_error, median_absolute_error

y_te = test_df["delay_minutes"]

med = train_df.groupby("train_number", observed=True)["delay_minutes"].median()
pred1 = test_df["train_number"].map(med).astype(float).fillna(train_df["delay_minutes"].median())
pred2 = test_df["previous_station_delay_min"].fillna(pred1)

print("Baseline 1 (train median)   MAE:", round(mean_absolute_error(y_te, pred1), 2),
      "| Median AE:", round(median_absolute_error(y_te, pred1), 2))
print("Baseline 2 (previous delay) MAE:", round(mean_absolute_error(y_te, pred2), 2),
      "| Median AE:", round(median_absolute_error(y_te, pred2), 2))
```

```python
keys = ["train_number", "run_date"]
g = df.groupby(keys, observed=True)

df["delay_2_stops_back_min"] = g["delay_minutes"].shift(2)
df["delay_change_last_stop_min"] = df["previous_station_delay_min"] - df["delay_2_stops_back_min"]

# pehle stop par NaN, taaki apni hi delay leak na ho
df["origin_delay_min"] = g["delay_minutes"].transform("first").where(df["stop_index"] > 0)

gp = df.groupby(keys, observed=True)["previous_station_delay_min"]
df["max_delay_so_far_min"] = gp.cummax()
df["avg_delay_so_far_min"] = gp.cumsum() / df["stop_index"].replace(0, float("nan"))

new_cols = ["previous_station_delay_min", "delay_2_stops_back_min", "delay_change_last_stop_min",
            "origin_delay_min", "max_delay_so_far_min", "avg_delay_so_far_min"]
print(df[["train_number", "run_date", "station_code", "delay_minutes"] + new_cols].head(8))
print(df[new_cols].isna().mean().round(4))
print(df[new_cols + ["delay_minutes"]].corr()["delay_minutes"].round(3))
```

```python
train_df = df[df["run_date"] < "2025-11-01"].copy()
val_df   = df[(df["run_date"] >= "2025-11-01") & (df["run_date"] < "2026-01-01")].copy()
test_df  = df[df["run_date"] >= "2026-01-01"].copy()

for name, d in [("Train", train_df), ("Val", val_df), ("Test", test_df)]:
    print(name, d.shape, d["run_date"].min().date(), "to", d["run_date"].max().date())
```

```python
train_avg = train_df.groupby("train_number", observed=True)["delay_minutes"].mean()
station_avg = train_df.groupby("station_code", observed=True)["delay_minutes"].mean()

for d in (train_df, val_df, test_df):
    d["train_avg_delay_min"] = d["train_number"].map(train_avg).astype(float)
    d["station_avg_delay_min"] = d["station_code"].map(station_avg).astype(float)

cols = ["train_avg_delay_min", "station_avg_delay_min"]
for name, d in [("Train", train_df), ("Val", val_df), ("Test", test_df)]:
    print(name, "NaN %:", (d[cols].isna().mean() * 100).round(2).to_dict())

print(train_df[cols + ["delay_minutes"]].corr()["delay_minutes"].round(3))
```

```python
import lightgbm as lgb
from sklearn.metrics import mean_absolute_error, median_absolute_error

# train_type ko sabhi hisso me same categories ke saath category banao
cats = sorted(df["train_type"].dropna().astype(str).unique())
for d in (train_df, val_df, test_df):
    d["train_type"] = pd.Categorical(d["train_type"].astype(str), categories=cats)

live_features = [
    "previous_station_delay_min", "delay_2_stops_back_min", "delay_change_last_stop_min",
    "origin_delay_min", "max_delay_so_far_min", "avg_delay_so_far_min",
    "avg_delay_last_7_runs_min", "train_avg_delay_min", "station_avg_delay_min",
    "day_of_week", "distance_from_origin_km", "scheduled_arrival_hour",
    "stop_index", "total_stops_in_route", "stops_remaining", "route_progress_ratio",
    "has_schedule_match", "train_type",
]

X_tr, y_tr = train_df[live_features], train_df["delay_minutes"]
X_va, y_va = val_df[live_features],   val_df["delay_minutes"]

live_model = lgb.LGBMRegressor(
    objective="regression_l1", n_estimators=2000, learning_rate=0.05,
    num_leaves=63, min_child_samples=50, subsample=0.8, subsample_freq=1,
    colsample_bytree=0.8, random_state=42, n_jobs=-1)

live_model.fit(X_tr, y_tr, eval_set=[(X_va, y_va)], eval_metric="l1",
               callbacks=[lgb.early_stopping(50), lgb.log_evaluation(100)])

pred_live = live_model.predict(X_va)
print("Best iteration:", live_model.best_iteration_)
print("LIVE model on VAL  MAE:", round(mean_absolute_error(y_va, pred_live), 2),
      "| Median AE:", round(median_absolute_error(y_va, pred_live), 2))

# Same val data par baselines, taaki fair comparison ho
med = train_df.groupby("train_number", observed=True)["delay_minutes"].median()
b1 = val_df["train_number"].map(med).astype(float).fillna(train_df["delay_minutes"].median())
b2 = val_df["previous_station_delay_min"].fillna(b1)
print("Baseline 1 on VAL  MAE:", round(mean_absolute_error(y_va, b1), 2))
print("Baseline 2 on VAL  MAE:", round(mean_absolute_error(y_va, b2), 2))
```

```python
pre_features = [
    "avg_delay_last_7_runs_min", "train_avg_delay_min", "station_avg_delay_min",
    "day_of_week", "distance_from_origin_km", "scheduled_arrival_hour",
    "stop_index", "total_stops_in_route", "stops_remaining", "route_progress_ratio",
    "has_schedule_match", "train_type",
]

pre_model = lgb.LGBMRegressor(
    objective="regression_l1", n_estimators=1500, learning_rate=0.08,
    num_leaves=63, min_child_samples=50, subsample=0.8, subsample_freq=1,
    colsample_bytree=0.8, random_state=42, n_jobs=-1)

pre_model.fit(train_df[pre_features], train_df["delay_minutes"],
              eval_set=[(val_df[pre_features], val_df["delay_minutes"])],
              eval_metric="l1",
              callbacks=[lgb.early_stopping(50), lgb.log_evaluation(200)])

pred_pre = pre_model.predict(val_df[pre_features])
print("Best iteration:", pre_model.best_iteration_)
print("PRE-DEPARTURE model on VAL  MAE:", round(mean_absolute_error(y_va, pred_pre), 2),
      "| Median AE:", round(median_absolute_error(y_va, pred_pre), 2))
print("Baseline 1 on VAL  MAE:", round(mean_absolute_error(y_va, b1), 2))
```

```python
import requests, re

def clean_name(n):
    n = str(n).upper()
    n = re.sub(r"\b(JN|JUNCTION|CANTT|CANT|CITY|TERMINUS|HALT|RLY|STN|STATION)\b\.?", "", n)
    return n.strip().title()

def geocode(name):
    try:
        r = requests.get("https://geocoding-api.open-meteo.com/v1/search",
                         params={"name": name, "count": 1, "country_code": "IN"}, timeout=10)
        res = r.json().get("results")
        if res:
            x = res[0]
            return x.get("latitude"), x.get("longitude"), x.get("admin1")
    except Exception as e:
        print("error:", name, e)
    return None, None, None

full = stations[["station_name", "station_full_name"]].drop_duplicates("station_name")
test_codes = ["GGC", "BTE", "NZM", "GKP", "LKO", "UMB", "CDG", "QG"]
test = full[full["station_name"].isin(test_codes)].copy()
test["query"] = test["station_full_name"].apply(clean_name)
test[["lat", "lon", "state"]] = test["query"].apply(lambda q: pd.Series(geocode(q)))
print(test[["station_name", "station_full_name", "query", "lat", "lon", "state"]])
```
